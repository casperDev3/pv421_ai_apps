## Вибір дому в **Хогвартсі**:

*   Грифіндор
*   Слизарин
*   Грифіндор
*   Слизарин




```python
import random
import time
from IPython.display import display, HTML

def sorting_hat_ultra():
    display(HTML("<h2>🧙‍♂️ Сортування починається...</h2>"))
    
    # ASCII Art
    hat_art = """
           ____
          /    \\
         |      |
         |______| 
        /        \\
       /          \\
      /____________\\
    """
    print(hat_art)
    time.sleep(1)

    phrases = [
        "\"Ооо... Бачу багато цікавого...\"",
        "\"Де ж твоє місце? Можливо, тут?..\"",
        "\"Ти маєш талант, о так...\"",
        "\"Цікаво... Дуже цікаво...\""
    ]

    print("--- ПИТАННЯ КАПЕЛЮХА ---")
    print("Яка цінність для вас найважливіша?")
    print("1. Сміливість (🦁)")
    print("2. Велич (🐍)")
    print("3. Знання (🦅)")
    print("4. Дружба (🦡)")

    choice = input("\nВаш вибір (1-4): ")

    for phrase in random.sample(phrases, 2):
        print(f"\n{phrase}")
        time.sleep(1.5)

    data = {
        "1": ("Грифіндор", "#740001", "🦁"),
        "2": ("Слизарин", "#1a472a", "🐍"),
        "3": ("Рейвенклов", "#0e1a40", "🦅"),
        "4": ("Гафлпаф", "#ecb939", "🦡")
    }

    houses = ["Грифіндор", "Слизарин", "Рейвенклов", "Гафлпаф"]
    weights = [0.1, 0.1, 0.1, 0.1]
    
    if choice in data:
        weights[int(choice)-1] = 0.85
    else:
        weights = [0.25, 0.25, 0.25, 0.25]

    result_name = random.choices(houses, weights=weights, k=1)[0]
    house_key = [k for k, v in data.items() if v[0] == result_name][0]
    name, color, icon = data[house_key]

    print("\"Я ВИРІШИВ!\"")
    time.sleep(1.5)

    display(HTML(f"""
        <div style='padding: 30px; border: 5px solid gold; border-radius: 15px; background-color: {color}; color: white; text-align: center; font-family: "Crimson Text", serif;'>
            <h1 style='font-size: 4em; margin: 0;'>{icon} {name.upper()} {icon}</h1>
            <p style='font-size: 1.5em;'>Твій шлях починається саме тут!</p>
        </div>
    """))

sorting_hat_ultra()
```
